Fill missing values using mean

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Matches_Played'] = df['Matches_Played'].fillna(df['Matches_Played'].mean())
df['Runs_Scored'] = df['Runs_Scored'].fillna(df['Runs_Scored'].mean())
df['Wickets_Taken'] = df['Wickets_Taken'].fillna(df['Wickets_Taken'].mean())
```

Check whether missing values are filled

```python
print(df.isnull().sum())
```

```python
df['Bowling_Style'] = df['Bowling_Style'].fillna(0)
```

Convert Categorical Columns into Numerical Values

```python
df = pd.get_dummies(df, columns=["Batting_Style", "Bowling_Style"])
```

```python
#Display the result
print(df.head())
```

Data Visualization

1. Histogram - Age Distribution

```python
df['Age'].plot(kind='hist', bins=10, edgecolor='black')
plt.title('Age Distribution of Cricket Players')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

2. Boxplot — Runs scored

```python
sns.boxplot(x=df['Runs_Scored'])
plt.title('Distribution of Runs Scored')
plt.show()
```

3. Scatter plot — Matches vs runs

```python
plt.scatter(df['Matches_Played'], df['Runs_Scored'])
plt.title('Matches Played vs Runs Scored')
plt.xlabel('Matches Played')
plt.ylabel('Runs Scored')
plt.show()
```

4. Heatmap - Correlation

```python
sns.heatmap(
    df.select_dtypes(include='number').corr(),
    annot=True,
    cmap='coolwarm'
)
plt.title('Correlation Heatmap')
plt.show()
```

5. BarChart - Players Role

```python
df['Role'].value_counts().plot(kind='bar', edgecolor='black')
plt.title('Number of Players by Role')
plt.xlabel('Role')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

6. Piechart - Batting Style

```python
# Since get_dummies split 'Batting_Style' into dummy columns, we sum those columns to get the distribution
batting_styles = df[['Batting_Style_Left-hand bat', 'Batting_Style_Right-hand bat']].sum()
batting_styles.index = ['Left-hand bat', 'Right-hand bat']

batting_styles.plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title('Batting Style Distribution')
plt.ylabel('')
plt.show()
```
